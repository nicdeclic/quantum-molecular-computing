# About The Author

Nicolas Joyeux

I am an independent researcher pursuing a new paradigm in quantum
information: computing built directly upon the physics of conjugated
carbon bonds.

I graduated with a B.Sc. in Computer Science from the Université de
Montréal in 1997. Over the decades, my path has taken me through the
complete stack of computation—from discrete transistor electronics and
microprocessor design up to low-level assembly and modern high-level
languages.

Investigating carbon-based quantum logic naturally forced me outside
traditional silos, driving me into chemistry, biology, mathematics, and
physics. While I feel I have only scratched the surface, this journey
has led me to erase more and more of the boundaries between disciplines.

Beyond the terminal and the lab bench, my thinking is anchored in the
philosophical foundations of deep ecology. I actively explore radical
alternatives for future societies — questioning our relationship with
the living world, re-evaluating technological necessity, and seeking
systemic pathways to genuinely improve how we inhabit this Earth.
